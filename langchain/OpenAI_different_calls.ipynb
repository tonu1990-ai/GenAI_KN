```python
import os
from dotenv import load_dotenv
load_dotenv()

OPENAPI_API_KEY = os.getenv("OPENAI_API_KEY")
```

```python
from openai import OpenAI
llm = OpenAI()
```

```python
# OpenAI call 1 

my_prompt = "Hi GPT I am Tonu James"
my_message = [{"role":"user", "content":my_prompt}]

response = llm.chat.completions.create(model='gpt-5-nano', messages = my_message )
response.choices[0].message.content
```

```python
# OpenAI call 2 - LangChain version
from langchain_openai import ChatOpenAI

# we use LLM gpt-4o 
llm2 = ChatOpenAI(model="gpt-5-nano")
my_prompt = "Hi GPT I am Tonu James"

response = llm2.invoke(my_prompt)

print(response.content)
```

```python
# LangChain version - using  message list
 

my_message = [
    ("human", "Hi GPT I am Tonu James")
]

response = llm2.invoke(my_message)

print(response.content)

 
```

```python
# another way - dictionaries 
my_message = [
    {"role": "user", "content": "Hi GPT I am Tonu James"}
]

response = llm2.invoke(my_message)

print(response.content)
```

```python
# Another approach from sesion 42 

from litellm import completion

response = completion(model="openai/gpt-5-nano" ,messages = my_message)
reply = response.choices[0].message.content
print(reply)
```
