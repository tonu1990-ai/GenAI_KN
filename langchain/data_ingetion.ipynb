```python
# Text loader

from langchain_community.document_loaders import TextLoader

loader_txt = TextLoader('transformer.txt')
loader_txt
```

```python
text_document = loader_txt.load()
print(text_document)
```

```python
type(text_document)
```

```python
type(text_document[0])
```

Now you can see the text type is Document. Remember we have to get our text in this forma for further processing.

Next step we perform the text splitig - the library used RecursiveCharacterTextSplitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=20)

# The below text_splitter.create_documents will give error as the text_document
# is already of type Document
# split_docs = text_splitter.create_documents(text_document)


split_docs = text_splitter.split_documents(text_document)
print(split_docs[0])
print(split_docs[1])
print(split_docs[2])
```
